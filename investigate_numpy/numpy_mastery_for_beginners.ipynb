# 🧠 01. NumPy Array Foundations & Memory Mechanics (Noob-Friendly Complete Guide)

* **Python List:** A loose plastic bag holding random object references. Slow & memory-heavy.
* **NumPy Array (`ndarray`):** An **egg carton** where every slot is identical in size and laid out consecutively in RAM. Fast & CPU-cache friendly!

## 📁 Section 6: Memory Views vs. Deep Copies


### 🔹 6.1: Slicing Shares Memory (Zero-Copy View)
**💡 What is it?** Slices point directly to the original array's memory buffer without copying.
**⚠️ Critical Note:** Mutating the slice mutates the parent array!


```python
original_account_balances = np.array([100.0, 250.0, 300.0, 500.0])
window_view = original_account_balances[0:2]
window_view[0] = 999.0
print('Original Modified via View!:', original_account_balances)
```

### 🔹 6.2: Explicit Deep Copy with .copy()
**💡 What is it?** Allocates completely fresh RAM for an independent duplicate.
**🤔 Why use it?** Safely altering a sub-array without risking changes to the raw source data.


```python
original_account_balances = np.array([100.0, 250.0, 300.0, 500.0])
safe_clone = original_account_balances[0:2].copy()
safe_clone[0] = 999.0
print('Original Untouched:', original_account_balances)
print('Safe Clone Mutated:', safe_clone)
```

**💡 What is it?** Direct item selection via `matrix[row_idx, col_idx]`.
**🤔 Why use it?** Accessing a specific customer feature or matrix entry.


```python
portfolio_matrix = np.array([
    [100.0, 105.0, 110.0],
    [200.0, 205.0, 210.0],
    [300.0, 310.0, 320.0]
])
print('Value at [Row 1, Col 2] :', portfolio_matrix[1, 2])
print('Entire Row 0 (Asset 1)  :', portfolio_matrix[0, :])
print('Entire Column 1 (Day 2) :', portfolio_matrix[:, 1])
```

## 📁 Section 2: Boolean Masking (Conditional Filtering)


### 🔹 2.1: Single Condition Filtering
**💡 What is it?** Applying a boolean filter expression inside brackets `arr[arr > threshold]`.
**🤔 Why use it?** Filtering valid transactions, isolating high-risk loans.


```python
customer_credit_scores = np.array([450, 820, 910, 550, 730, 600, 380])
prime_credit_scores = customer_credit_scores[customer_credit_scores >= 700]
print('Prime Credit Scores (>= 700):', prime_credit_scores)
```

### 🔹 2.2: Compound Conditions with Bitwise & / |
**💡 What is it?** Combining multiple criteria using bitwise `&` (AND) and `|` (OR) with parentheses `()`.
**⚠️ Critical Note:** Standard Python `and` / `or` fails on NumPy arrays; you MUST use `&` / `|`.


```python
customer_credit_scores = np.array([450, 820, 910, 550, 730, 600, 380])
near_prime_scores = customer_credit_scores[(customer_credit_scores >= 600) & (customer_credit_scores < 750)]
print('Near-Prime Scores (600 to 749):', near_prime_scores)
```

### 🔹 2.3: In-Place Conditional Mutation
**💡 What is it?** Overwriting only the values that satisfy a condition.
**🤔 Why use it?** Winsorizing outliers, clamping negative balances, risk capping.


```python
account_balances = np.array([-50.0, 1200.0, -15.0, 3400.0, 500.0])
account_balances[account_balances < 0] = 0.0
print('Negative Balances Clamped to Zero:', account_balances)
```

## 📁 Section 3: np.where (Vectorized Ternary Decisions)


### 🔹 3.1: Vectorized If-Else with np.where()
**💡 What is it?** `np.where(condition, value_if_true, value_if_false)`.
**🤔 Why use it?** Vectorized `CASE WHEN` logic evaluated across millions of elements in parallel.


```python
account_balances = np.array([1200.0, -50.0, 3400.0, -200.0, 500.0])
account_risk_flags = np.where(account_balances >= 0, 'Solvent', 'Overdrawn')
print('Account Solvency Statuses:', account_risk_flags)
```

### 🔹 3.2: Coordinate Extraction with np.where()
**💡 What is it?** Calling `np.where(condition)` without replacement returns indices matching condition.
**🤔 Why use it?** Finding exact row positions of flagged transactions.


```python
account_balances = np.array([1200.0, -50.0, 3400.0, -200.0, 500.0])
overdrawn_indices = np.where(account_balances < 0)[0]
print('Indices of Overdrawn Accounts:', overdrawn_indices)
```

## 📁 Section 1: Global vs Directional Reductions


### 🔹 1.1: Global Aggregations Across Entire Array
**💡 What is it?** Reduces a 2D matrix to a single summary number.
**🤔 Why use it?** Calculating total portfolio value or overall volatility.


```python
financial_matrix = np.array([
    [10.0, 20.0, 30.0],
    [40.0, 50.0, 60.0]
], dtype=np.float64)
print('Global Sum :', financial_matrix.sum())
print('Global Mean:', financial_matrix.mean())
print('Global Max :', financial_matrix.max())
print('Global Std :', financial_matrix.std())
```

**💡 What is it?** Sums down rows to yield 1 summary value per column.

```python
financial_matrix = np.array([
    [10.0, 20.0, 30.0],
    [40.0, 50.0, 60.0]
], dtype=np.float64)
feature_means = financial_matrix.mean(axis=0)
print('Feature Column Means (axis=0):', feature_means)  # Shape (3,)
```

## 📁 Section 2: Preserving Dimensions with keepdims=True


### 🔹 2.1: Broadcasting Safety with keepdims=True
**💡 What is it?** Keeps reduced axis as size `1` (shape `(2, 1)` instead of collapsing to `(2,)`).
**🤔 Why use it?** Enables direct subtraction from original `(2, 3)` matrix without shape alignment bugs.


```python
financial_matrix = np.array([
    [10.0, 20.0, 30.0],
    [40.0, 50.0, 60.0]
], dtype=np.float64)
row_means_aligned = financial_matrix.mean(axis=1, keepdims=True)
print('Row Means Shape (with keepdims):', row_means_aligned.shape)
mean_centered_matrix = financial_matrix - row_means_aligned
print('Mean-Centered Matrix:\n', mean_centered_matrix)
```

# 🧠 04. Broadcasting Rules & Memory Mechanics (Noob-Friendly Complete Guide)

### 💡 Mental Model: The Magic Virtual Grid Stretcher
When performing math between arrays of different shapes, NumPy **virtually stretches** the smaller array across missing dimensions with **ZERO RAM duplication**!
### 📐 The 2 Golden Rules of Broadcasting:
Comparing shapes **from right to left (trailing dimensions)**:
1. Dimensions are **strictly equal**, OR
2. One of the dimensions is **1** (which stretches to match the larger dimension).

## 📁 Section 1: Fundamental Broadcasting Patterns


### 🔹 1.1: Scalar to Multi-Dimensional Array
**💡 What is it?** Applying an operation between a single scalar and an entire matrix.
**🤔 Why use it?** Adding fixed fees, scaling currency units.


```python
asset_prices = np.array([[10.0, 20.0], [30.0, 40.0]])
print('Add Flat Trading Fee (+$5):\n', asset_prices + 5.0)
print('Apply 2x Leverage (* 2)   :\n', asset_prices * 2.0)
```

### 🔹 1.2: 1D Row Vector + 2D Matrix (Trailing Alignment)
**💡 What is it?** Adding shape `(3,)` to `(2, 3)` stretches the row vector across all rows.
**🤔 Why use it?** Applying category-specific baseline adjustments to each user's spend profile.


```python
user_spending_matrix = np.array([
    [100.0, 200.0, 300.0],
    [400.0, 500.0, 600.0]
])  # Shape (2, 3)
category_adjustments = np.array([5.0, 10.0, 15.0])  # Shape (3,)
print('Adjusted Spend Matrix:\n', user_spending_matrix + category_adjustments)
```

### 🔹 1.3: Column Vector Alignment with np.newaxis / [:, None]
**💡 What is it?** Reshaping a 1D vector `(2,)` into `(2, 1)` so it stretches horizontally across all columns.
**🤔 Why use it?** Subtracting each individual user's mean balance across all their accounts.


```python
user_spending_matrix = np.array([
    [100.0, 200.0, 300.0],
    [400.0, 500.0, 600.0]
])  # Shape (2, 3)
user_bonuses = np.array([50.0, 100.0])[:, None]  # Shape (2, 1)
print('Column Vector Shape:', user_bonuses.shape)
print('Column-Stretched Result:\n', user_spending_matrix + user_bonuses)
```

### 🔹 2.1: Generating 2D Grids from Two 1D Vectors

```python
interest_rates = np.array([0.03, 0.05, 0.07])[:, None]  # Shape (3, 1)
principal_amounts = np.array([1000, 5000, 10000, 50000]) # Shape (4,)
annual_interest_grid = interest_rates * principal_amounts # Shape (3, 4)
print('Annual Interest Payoff Grid (3x4):\n', annual_interest_grid)
```

## 📁 Section 3: Real-World Fintech: Feature Z-Score Standardization


### 🔹 3.1: Standardizing Data with Broadcasting (Mean=0, Std=1)
**💡 What is it?** $Z = \frac{X - \mu}{\sigma}$. Centers features to zero mean and unit variance.
**🤔 Why use it?** Standardizing tabular credit underwriting features for machine learning models.


```python
raw_features = np.array([[100.0, 15.0], [200.0, 30.0], [300.0, 45.0]])
feature_means = raw_features.mean(axis=0)  # Shape (2,)
feature_stds = raw_features.std(axis=0)    # Shape (2,)
standardized_features = (raw_features - feature_means) / feature_stds
print('Standardized Features (Z-Scores):\n', standardized_features)
```

# 🧠 12. Compiled Code with Numba (`@njit`) (Noob-Friendly Complete Guide)

### 💡 Mental Model: Python Loops Running at C Speed
Python `for` loops are slow because the interpreter checks variable types on every step.
**Numba (`@njit`)** compiles your Python function into **machine code (LLVM)** on the fly, giving you **100x to 500x speedups** with zero C code!

---
### 📌 How this notebook is structured:
1. **Self-Documenting Names:** No single-letter variables (`m`, `v`, `a`, `x`). Every variable explains its exact role (e.g. `transaction_amounts`, `user_feature_matrix`).

---
## 📁 Section 1: Basic JIT Compilation


### 🔹 1.1: Accelerating Loops with @njit
**💡 What is it?** Decorating a function with `@njit` compiles it to bare-metal machine instructions.
**🤔 Why use it?** Eliminates loop overhead in sequential algorithms (e.g. Monte Carlo, path simulation).


```python
from numba import njit

@njit
def fast_monte_carlo_accumulator(num_iterations):
    running_total = 0.0
    for step_idx in range(num_iterations):
        running_total += step_idx * 1.5
    return running_total

print('Compiled Fast Accumulator Result:', fast_monte_carlo_accumulator(1000000))
```

### 🔹 1.2: JIT Compiling Custom Array Filters
**💡 What is it?** Compiling functions that mutate and iterate over NumPy arrays.
**🤔 Why use it?** Complex custom algorithms that cannot be easily vectorized.


```python
from numba import njit

@njit
def exponential_decay_filter(price_series, decay_rate):
    smoothed_output = np.empty_like(price_series)
    smoothed_output[0] = price_series[0]
    for i in range(1, len(price_series)):
        smoothed_output[i] = decay_rate * price_series[i] + (1 - decay_rate) * smoothed_output[i-1]
    return smoothed_output

raw_prices = np.array([100.0, 102.0, 101.0, 105.0, 108.0])
print('Custom Smoothed Output:', exponential_decay_filter(raw_prices, 0.5))
```

---
## 📁 Section 2: Parallel Multi-Core Execution with prange


### 🔹 2.1: Multi-Threaded Loops with parallel=True & prange()
**💡 What is it?** Automatically spreads loop iterations across all CPU cores.
**🤔 Why use it?** Massive parallel simulations across millions of iterations.


```python
from numba import njit, prange

@njit(parallel=True)
def parallel_sample_counter(num_samples):
    total_hits = 0
    for sample_idx in prange(num_samples):
        total_hits += (sample_idx % 2)
    return total_hits

print('Parallel Execution Total Hits:', parallel_sample_counter(5000000))
```
