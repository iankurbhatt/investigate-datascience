# 🧠 10. Structured & Record Arrays (C-Structs in NumPy) (Noob-Friendly Complete Guide)

### 💡 Mental Model: A SQL Table in a Single NumPy Array
Structured arrays let you define **heterogeneous typed columns** (e.g. `int`, `float`, `string`) stored side-by-side like a C-`struct` in memory.

---
### 📌 How this notebook is structured:
1. **Self-Documenting Names:** No single-letter variables (`m`, `v`, `a`, `x`). Every variable explains its exact role (e.g. `transaction_amounts`, `user_feature_matrix`).
2. **Atomic Concept Pairs:** Every concept has an isolated **Markdown explanation (What, Why, Mental Model)** followed immediately by a **short 3–5 line code cell**.

```python
import numpy as np
print(f"🚀 NumPy {np.__version__} Ready!")
```

---
## 📁 Section 1: Defining Structured Schemas


### 🔹 1.1: Creating Structured Dtype Schema
**💡 What is it?** Defining named fields and data types: `('customer_id', 'i4'), ('account_balance', 'f8'), ('tier', 'U10')`.
**🤔 Why use it?** Storing tabular records in ultra-compact binary memory.


```python
customer_record_schema = np.dtype([
    ('customer_id', np.int32),
    ('account_balance', np.float64),
    ('tier', 'U10')
])
print('Custom Customer Dtype Schema:', customer_record_schema)
```

### 🔹 1.2: Populating Structured Records
**💡 What is it?** Creating an array matching the custom schema.
**🤔 Why use it?** High-speed tabular processing without Pandas overhead.


```python
customer_records_table = np.array([
    (101, 1540.50, 'Gold'),
    (102, 450.00, 'Silver'),
    (103, 8900.25, 'Platinum')
], dtype=customer_record_schema)
print('Structured Customer Table:\n', customer_records_table)
```

---
## 📁 Section 2: Querying Structured Fields & Records


### 🔹 2.1: Accessing Specific Column by Field Name
**💡 What is it?** `records['account_balance']` extracts an entire column vector.
**🤔 Why use it?** Vectorized operations on single table columns.


```python
print('All Customer Balances :', customer_records_table['account_balance'])
print('Average Account Balance: $', customer_records_table['account_balance'].mean())
```

### 🔹 2.2: Filtering Records by Condition
**💡 What is it?** Applying boolean masks directly to structured fields.
**🤔 Why use it?** Filtering table rows at C-speed.


```python
high_value_customers = customer_records_table[customer_records_table['account_balance'] > 1000.0]
print('High-Value Accounts:\n', high_value_customers)
```

---
## 📁 Section 3: Record Arrays (Dot-Access Syntax)


### 🔹 3.1: np.recarray for Dot-Notation Access
**💡 What is it?** Wraps structured array so fields can be accessed via `rec.account_balance` instead of `rec['account_balance']`.
**🤔 Why use it?** Cleaner object-like syntax.


```python
customer_recarray = customer_records_table.view(np.recarray)
print('Customer IDs via Dot Notation:', customer_recarray.customer_id)
print('Tiers via Dot Notation       :', customer_recarray.tier)
```
