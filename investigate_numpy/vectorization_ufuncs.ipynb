## 📁 Section 1: Math & Transcendental Vector Functions


### 🔹 1.1: Element-Wise Arithmetic & Roots
**💡 What is it?** High-speed SIMD math executed across entire arrays.
**🤔 Why use it?** Blazing fast math without for-loops.


```python
sample_values = np.array([1.0, 4.0, 9.0, 16.0])
print('Square Roots :', np.sqrt(sample_values))
print('Squares      :', np.square(sample_values))
print('Reciprocals  :', np.reciprocal(sample_values))
```

### 🔹 1.2: Logarithmic & Exponential Transforms
**💡 What is it?** `np.exp()` and `np.log1p()` for stable logarithmic and exponential scaling.
**🤔 Why use it?** Computing log-returns, softmax activations, exponential risk decay.


```python
daily_percentage_returns = np.array([0.01, 0.05, -0.02, 0.10])
log_returns = np.log1p(daily_percentage_returns)
print('Numerically Stable Log Returns:', log_returns)
```

### 🔹 1.3: Trigonometric Transforms
**💡 What is it?** `np.sin()`, `np.cos()`, `np.arctan()`.
**🤔 Why use it?** Signal processing, Fourier cycles, seasonal wave modeling.


```python
cycle_angles = np.array([0, np.pi/4, np.pi/2, np.pi])
print('Sine Wave Outputs:', np.round(np.sin(cycle_angles), 4))
```

## 📁 Section 2: Advanced Ufunc Methods (.reduce, .accumulate, .outer, .at)


### 🔹 2.1: Custom Reductions with ufunc.reduce()
**💡 What is it?** Repeatedly applies any binary ufunc until one summary value remains.
**🤔 Why use it?** Universal reducer across any custom mathematical operator.


```python
numbers_sequence = np.array([1, 2, 3, 4, 5])
print('np.add.reduce (Total Sum)     :', np.add.reduce(numbers_sequence))
print('np.multiply.reduce (Factorial):', np.multiply.reduce(numbers_sequence))
```

### 🔹 2.2: Running Accumulations with ufunc.accumulate()
**💡 What is it?** Captures intermediate reduction values as an ongoing running vector.
**🤔 Why use it?** Tracking portfolio high-water marks and maximum drawdowns.


```python
stock_price_history = np.array([100.0, 105.0, 98.0, 112.0, 108.0])
running_peak_price = np.maximum.accumulate(stock_price_history)
print('Running High-Water Mark Peak:', running_peak_price)
```

### 🔹 2.3: Pairwise Combinations with ufunc.outer()
**💡 What is it?** Applies operation to all pairs $(a_i, b_j)$.
**🤔 Why use it?** Pairwise difference matrices, distance metrics.


```python
bid_prices = np.array([100.0, 101.0, 102.0])
ask_prices = np.array([103.0, 104.0, 105.0])
spread_matrix = np.subtract.outer(ask_prices, bid_prices)
print('Pairwise Spread Grid (Ask - Bid):\n', spread_matrix)
```

### 🔹 2.4: Unbuffered In-Place Updates with ufunc.at()
**💡 What is it?** Performs in-place unbuffered updates even when indices repeat.
**🤔 Why use it?** Incrementing histogram bins and transaction counter buckets.


```python
event_frequency_bins = np.zeros(5, dtype=np.int32)
recorded_event_indices = np.array([1, 1, 2, 1, 4])  # Index 1 repeats 3 times
np.add.at(event_frequency_bins, recorded_event_indices, 10)
print('Aggregated Event Frequency Bins:', event_frequency_bins)
```

## 📁 Section 3: Custom Vectorized Functions with np.vectorize


### 🔹 3.1: Custom Business Logic Vectorization
**💡 What is it?** Converts standard Python if-else logic into a vectorized array function.
**🤔 Why use it?** Applying complex underwriting decision trees to array columns cleanly.


```python
def evaluate_credit_tier(fico, dti):
    return 'Approved' if fico >= 700 and dti < 0.40 else 'Declined'
vectorized_evaluator = np.vectorize(evaluate_credit_tier)
customer_ficos = np.array([720, 680, 750, 620])
customer_dtis = np.array([0.35, 0.20, 0.45, 0.50])
print('Underwriting Decisions:', vectorized_evaluator(customer_ficos, customer_dtis))
```
