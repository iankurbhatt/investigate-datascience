## 📁 Section 1: Numeric Data Types & Memory Footprint


### 🔹 1.1: Comparing Integer Types (int8 vs int64)
**💡 What is it?** Integer types with explicit bit widths.
**🤔 Why use it?** `int8` holds $[-128, 127]$ in 1 byte; `int64` takes 8 bytes per number.


```python
small_flags_int8 = np.array([10, 20, 120], dtype=np.int8)
large_integers_int64 = np.array([10, 20, 120], dtype=np.int64)
print('int8 Total Bytes (3 slots) : ', small_flags_int8.nbytes)
print('int64 Total Bytes (3 slots): ', large_integers_int64.nbytes)  # 8x larger!
```

### 🔹 1.2: Floating Point Precision (float32 vs float64)
**💡 What is it?** Trade-off between decimal precision and RAM size.
**🤔 Why use it?** `float32` is standard for Deep Learning and GPU training.


```python
single_precision_float = np.array([1.23456789], dtype=np.float32)
double_precision_float = np.array([1.23456789], dtype=np.float64)
print('float32 Precision (7 decimals) :', single_precision_float[0])
print('float64 Precision (16 decimals):', double_precision_float[0])
```

## 📁 Section 2: Type Casting & Overflow Limits


### 🔹 2.1: Explicit Type Casting with .astype()
**💡 What is it?** Converts an array to a new data type.
**🤔 Why use it?** Downcasting feature matrices before saving.


```python
raw_floating_prices = np.array([10.50, 20.80, 30.10])
truncated_integer_prices = raw_floating_prices.astype(np.int32)
print('Casted to int32 (Truncated):', truncated_integer_prices)
```

### 🔹 2.2: Inspecting Type Limits with np.iinfo & np.finfo
**💡 What is it?** Queries the exact minimum, maximum, and precision limits of any data type.
**🤔 Why use it?** Preventing numeric overflow / underflow bugs.


```python
print('int16 Safe Bounds:', np.iinfo(np.int16).min, 'to', np.iinfo(np.int16).max)
print('float32 Machine Epsilon:', np.finfo(np.float32).eps)
```
