# 🧠 03. Axis Aggregations & Dimensional Reduction (Noob-Friendly Complete Guide)

### 💡 Mental Model: The Direction of Collapse
When aggregating along an `axis`, that specific dimension is **COLLAPSED / REDUCED**!
* `axis=0` = Collapse across **ROWS** $\rightarrow$ Results in Column Statistics.
* `axis=1` = Collapse across **COLUMNS** $\rightarrow$ Results in Row Statistics.

## 📁 Section 1: Global vs Directional Reductions


### 🔹 1.1: Global Aggregations Across Entire Array
**💡 What is it?** Reduces a 2D matrix to a single summary number.
**🤔 Why use it?** Calculating total portfolio value or overall volatility.


```python
financial_matrix = np.array([
    [10.0, 20.0, 30.0],
    [40.0, 50.0, 60.0]
], dtype=np.float64)
print('Global Sum :', financial_matrix.sum())
print('Global Mean:', financial_matrix.mean())
print('Global Max :', financial_matrix.max())
print('Global Std :', financial_matrix.std())
```

### 🔹 1.2: Column-Wise Aggregation (axis=0)
**💡 What is it?** Sums down rows to yield 1 summary value per column.
**🤔 Why use it?** Finding the average value for each feature across all customers.


```python
financial_matrix = np.array([
    [10.0, 20.0, 30.0],
    [40.0, 50.0, 60.0]
], dtype=np.float64)
feature_means = financial_matrix.mean(axis=0)
print('Feature Column Means (axis=0):', feature_means)  # Shape (3,)
```

### 🔹 1.3: Row-Wise Aggregation (axis=1)
**💡 What is it?** Sums across columns to yield 1 summary value per row.
**🤔 Why use it?** Calculating each customer's total spending across all categories.


```python
financial_matrix = np.array([
    [10.0, 20.0, 30.0],
    [40.0, 50.0, 60.0]
], dtype=np.float64)
customer_total_spend = financial_matrix.sum(axis=1)
print('Customer Spend Totals (axis=1):', customer_total_spend)  # Shape (2,)
```

---
## 📁 Section 2: Preserving Dimensions with keepdims=True


### 🔹 2.1: Broadcasting Safety with keepdims=True
**💡 What is it?** Keeps reduced axis as size `1` (shape `(2, 1)` instead of collapsing to `(2,)`).
**🤔 Why use it?** Enables direct subtraction from original `(2, 3)` matrix without shape alignment bugs.


```python
financial_matrix = np.array([
    [10.0, 20.0, 30.0],
    [40.0, 50.0, 60.0]
], dtype=np.float64)
row_means_aligned = financial_matrix.mean(axis=1, keepdims=True)
print('Row Means Shape (with keepdims):', row_means_aligned.shape)
mean_centered_matrix = financial_matrix - row_means_aligned
print('Mean-Centered Matrix:\n', mean_centered_matrix)
```

## 📁 Section 3: Finding Coordinates of Peaks (argmin / argmax)


### 🔹 3.1: Finding Peak & Trough Indices
**💡 What is it?** Returns the exact index coordinate where the maximum or minimum value occurs.
**🤔 Why use it?** Finding the exact day of peak asset price or lowest system throughput.


```python
daily_stock_prices = np.array([102.5, 108.3, 97.4, 115.0, 110.2])
peak_day_index = np.argmax(daily_stock_prices)
print(f'Peak Price ${daily_stock_prices[peak_day_index]} occurred on Day Index {peak_day_index}')
```

---
## 📁 Section 4: Cumulative Running Sequences


### 🔹 4.1: Running Cumulative Sum with np.cumsum()
**💡 What is it?** Computes running cumulative totals element-by-element.
**🤔 Why use it?** Generating cumulative P&L equity curves, tracking running balances.


```python
daily_pnl_returns = np.array([100.0, -50.0, 200.0, 300.0, -150.0])
cumulative_equity_curve = np.cumsum(daily_pnl_returns)
print('Cumulative Equity Curve:', cumulative_equity_curve)
```

## 📁 Section 5: Robust Handling of Missing Data (NaN-Safe)


### 🔹 5.1: Safe Reductions with np.nanmean and np.nansum
**💡 What is it?** Standard `np.mean()` turns into `NaN` if any value is missing. `np.nanmean()` ignores missing values.
**🤔 Why use it?** Calculating statistics on production telemetry with sensor outages.


```python
sensor_telemetry = np.array([10.0, 20.0, np.nan, 40.0])
print('Standard np.mean (Polluted with NaN):', np.mean(sensor_telemetry))
print('NaN-Safe np.nanmean (Clean Average) :', np.nanmean(sensor_telemetry))
```
