## 📁 Section 1: Basic JIT Compilation


### 🔹 1.1: Accelerating Loops with @njit
**💡 What is it?** Decorating a function with `@njit` compiles it to bare-metal machine instructions.
**🤔 Why use it?** Eliminates loop overhead in sequential algorithms (e.g. Monte Carlo, path simulation).


```python
from numba import njit

@njit
def fast_monte_carlo_accumulator(num_iterations):
    running_total = 0.0
    for step_idx in range(num_iterations):
        running_total += step_idx * 1.5
    return running_total

print('Compiled Fast Accumulator Result:', fast_monte_carlo_accumulator(1000000))
```

### 🔹 1.2: JIT Compiling Custom Array Filters
**💡 What is it?** Compiling functions that mutate and iterate over NumPy arrays.
**🤔 Why use it?** Complex custom algorithms that cannot be easily vectorized.


```python
from numba import njit

@njit
def exponential_decay_filter(price_series, decay_rate):
    smoothed_output = np.empty_like(price_series)
    smoothed_output[0] = price_series[0]
    for i in range(1, len(price_series)):
        smoothed_output[i] = decay_rate * price_series[i] + (1 - decay_rate) * smoothed_output[i-1]
    return smoothed_output

raw_prices = np.array([100.0, 102.0, 101.0, 105.0, 108.0])
print('Custom Smoothed Output:', exponential_decay_filter(raw_prices, 0.5))
```

## 📁 Section 2: Parallel Multi-Core Execution with prange


### 🔹 2.1: Multi-Threaded Loops with parallel=True & prange()
**💡 What is it?** Automatically spreads loop iterations across all CPU cores.
**🤔 Why use it?** Massive parallel simulations across millions of iterations.


```python
from numba import njit, prange

@njit(parallel=True)
def parallel_sample_counter(num_samples):
    total_hits = 0
    for sample_idx in prange(num_samples):
        total_hits += (sample_idx % 2)
    return total_hits

print('Parallel Execution Total Hits:', parallel_sample_counter(5000000))
```
