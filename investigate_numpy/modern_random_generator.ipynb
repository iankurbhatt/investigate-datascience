## 📁 Section 1: Initializing Generator & Continuous Sampling


### 🔹 1.1: Spawning Reproducible Generator Instance
**💡 What is it?** `rng = np.random.default_rng(seed)` creates an isolated pseudo-random generator.
**🤔 Why use it?** Guarantees deterministic repeatability across test suites and ML runs.


```python
random_generator = np.random.default_rng(seed=42)
uniform_floats = random_generator.random((2, 3))
print('Uniform Random Floats [0.0, 1.0):\n', uniform_floats)
```

### 🔹 1.2: Generating Discrete Integers with rng.integers()
**💡 What is it?** Generates random integer values between `[low, high)`.
**🤔 Why use it?** Simulating dice rolls, synthetic customer IDs, discrete arrival events.


```python
simulated_dice_rolls = random_generator.integers(low=1, high=7, size=5)
print('5 Simulated Dice Rolls:', simulated_dice_rolls)
```

---
## 📁 Section 2: Statistical Distributions (Normal & Uniform)


### 🔹 2.1: Gaussian (Normal) Distribution with rng.normal()
**💡 What is it?** Bell-curve distribution centered at mean `loc` with spread `scale`.
**🤔 Why use it?** Simulating asset return shocks, measurement noise, ML initialization.


```python
daily_asset_returns = random_generator.normal(loc=0.001, scale=0.02, size=5)
print('Simulated Daily Asset Returns (Mean 0.1%, Vol 2%):\n', daily_asset_returns)
```

### 🔹 2.2: Bounded Uniform Distribution with rng.uniform()
**💡 What is it?** Equal probability of selecting any value between `[low, high)`.
**🤔 Why use it?** Monte Carlo risk simulations, physics coordinate sampling.


```python
simulated_temperatures = random_generator.uniform(low=20.0, high=35.0, size=4)
print('Simulated Temperature Readings:', simulated_temperatures)
```

---
## 📁 Section 3: Sampling, Permutations & Shuffling


### 🔹 3.1: Random Choice with / without Replacement
**💡 What is it?** Samples elements from an array with custom probabilities.
**🤔 Why use it?** Statistical bootstrapping, building random cross-validation folds.


```python
card_deck = np.array(['Ace', 'King', 'Queen', 'Jack', '10'])
sampled_hand = random_generator.choice(card_deck, size=3, replace=False)
print('Sampled Hand (No Duplicates Allowed):', sampled_hand)
```

### 🔹 3.2: Array Shuffling & Permutations
**💡 What is it?** `rng.permutation()` randomizes the ordering of elements.
**🤔 Why use it?** Shuffling training samples before neural network epochs.


```python
ordered_sample_ids = np.arange(1, 6)
shuffled_sample_ids = random_generator.permutation(ordered_sample_ids)
print('Original Sequence:', ordered_sample_ids)
print('Shuffled Sequence:', shuffled_sample_ids)
```
