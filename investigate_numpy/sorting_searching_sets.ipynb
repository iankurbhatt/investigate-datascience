## 📁 Section 1: Sorting Operations


### 🔹 1.1: Direct Sorting with np.sort()
**💡 What is it?** Orders numbers from smallest to largest.
**🤔 Why use it?** Ranking prices, computing percentiles.


```python
raw_scores = np.array([45, 12, 89, 34, 67])
print('Sorted Array Copy:', np.sort(raw_scores))
raw_scores.sort()  # In-place sort
print('Sorted In-Place  :', raw_scores)
```

### 🔹 1.2: Indirect Sorting with np.argsort()
**💡 What is it?** Returns the permutation of indices that sorts the array.
**🤔 Why use it?** Sorting parallel arrays (e.g. ordering user names by their account balance).


```python
customer_balances = np.array([4500.0, 1200.0, 8900.0, 3400.0])
customer_names = np.array(['Alice', 'Bob', 'Charlie', 'David'])
ranking_indices = np.argsort(customer_balances)
print('Sorted Balances:', customer_balances[ranking_indices])
print('Sorted Users   :', customer_names[ranking_indices])
```

---
## 📁 Section 2: Binary Search with np.searchsorted


### 🔹 2.1: Fast $O(\log N)$ Bucket Lookup with np.searchsorted()
**💡 What is it?** Finds where numbers fit inside sorted boundaries in logarithmic time.
**🤔 Why use it?** Assigning credit score bands, tax brackets, latency histogram bins.


```python
fico_tier_thresholds = np.array([580, 670, 740, 800])  # Must be sorted!
applicant_scores = np.array([620, 710, 850, 550, 780])
assigned_tier_ids = np.searchsorted(fico_tier_thresholds, applicant_scores)
print('Applicant Scores :', applicant_scores)
print('Assigned Tier IDs:', assigned_tier_ids)
```

---
## 📁 Section 3: Fast Partial Sorting (Top-K with np.partition)


### 🔹 3.1: Finding Top-K Elements in $O(N)$ Time with np.argpartition()
**💡 What is it?** Moves the $k$ smallest/largest elements without doing an expensive full sort.
**🤔 Why use it?** Finding the top 3 worst latency spikes out of 10,000,000 requests in milliseconds.


```python
api_latencies_ms = np.array([12, 45, 98, 23, 150, 87, 34, 210, 56])
worst_3_indices = np.argpartition(api_latencies_ms, -3)[-3:]
print('Top 3 Slowest Latencies (ms):', api_latencies_ms[worst_3_indices])
```

## 📁 Section 4: Set Operations & Uniqueness


### 🔹 4.1: Unique Elements & Frequencies with np.unique()
**💡 What is it?** Returns distinct values and their occurrence counts.
**🤔 Why use it?** Categorical distribution analysis, trade action breakdown.


```python
trade_actions = np.array(['BUY', 'SELL', 'BUY', 'HOLD', 'BUY', 'SELL'])
unique_actions, action_counts = np.unique(trade_actions, return_counts=True)
print('Unique Actions:', unique_actions)
print('Action Counts :', action_counts)
```

### 🔹 4.2: Set Intersections & Differences (np.intersect1d, np.setdiff1d)
**💡 What is it?** Mathematical set comparisons between two arrays.
**🤔 Why use it?** Identifying retained vs churned customers between consecutive months.


```python
january_active_users = np.array([101, 102, 103, 104, 105])
february_active_users = np.array([103, 104, 105, 106, 107])
print('Retained Users (Overlap):', np.intersect1d(january_active_users, february_active_users))
print('Churned Users  (Left Jan):', np.setdiff1d(january_active_users, february_active_users))
```

### 🔹 4.3: Membership Filtering with np.isin()
**💡 What is it?** Returns a boolean mask checking if elements exist inside a reference set.
**🤔 Why use it?** Vectorized SQL `WHERE id IN (...)` filtering.


```python
processed_transactions = np.array([101, 204, 305, 408, 512])
fraud_blacklist = np.array([204, 512, 999])
is_fraud_flagged = np.isin(processed_transactions, fraud_blacklist)
print('Fraud Mask Flag   :', is_fraud_flagged)
print('Flagged Txn IDs   :', processed_transactions[is_fraud_flagged])
```
