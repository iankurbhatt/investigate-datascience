### 💡 Mental Model: Reusing Memory vs Creating Garbage
* `a = a + b` $\rightarrow$ Allocates a brand new temporary array in RAM (slow!).
* `a += b` $\rightarrow$ Modifies `a` **in-place** with zero new memory allocation (blazing fast!).

## 📁 Section 1: In-Place Arithmetic Mechanics


### 🔹 1.1: In-Place Mutation (+=, *=) vs Out-of-Place (+)
**💡 What is it?** Reusing existing RAM addresses.
**🤔 Why use it?** Prevents Out-Of-Memory (OOM) crashes on large arrays.


```python
large_memory_buffer = np.ones((1000, 1000), dtype=np.float64)
address_before_mutation = large_memory_buffer.__array_interface__['data'][0]
large_memory_buffer += 5.0  # In-place addition
address_after_mutation = large_memory_buffer.__array_interface__['data'][0]
print('Memory Address Preserved (Zero New RAM Allocation)?:', address_before_mutation == address_after_mutation)
```

## 📁 Section 2: Inspecting Array Memory Flags


### 🔹 2.1: Checking Ownership & Contiguity Flags
**💡 What is it?** `arr.flags` reveals whether the array owns its memory buffer and if it's C-contiguous.
**🤔 Why use it?** Verifying if an operation triggered an unintended deep copy.


```python
parent_dataset = np.arange(10)
sliced_subview = parent_dataset[::2]
print('Parent Dataset Owns Data?:', parent_dataset.flags.owndata)  # True
print('Sliced Subview Owns Data?:', sliced_subview.flags.owndata) # False (Points to parent)
```
