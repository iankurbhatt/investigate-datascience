### 💡 Mental Model: Row-Major (C) vs Column-Major (Fortran)
* **C-Order (Row-Major):** Row elements are adjacent in RAM. Fast row iteration.
* **Fortran-Order (Column-Major):** Column elements are adjacent in RAM. Fast column iteration.

## 📁 Section 1: C-Order vs Fortran-Order


### 🔹 1.1: Comparing Memory Layouts (np.ascontiguousarray)
**💡 What is it?** Checking and converting array memory ordering.
**🤔 Why use it?** Ensuring optimal CPU cache locality for downstream libraries (C++/Fortran/CUDA).


```python
c_contiguous_grid = np.ones((3, 3), order='C')
fortran_contiguous_grid = np.ones((3, 3), order='F')
print('C-Array Strides (Row contiguous)   :', c_contiguous_grid.strides)
print('F-Array Strides (Column contiguous):', fortran_contiguous_grid.strides)
```

## 📁 Section 2: Flattening vs Ravelling


### 🔹 2.1: Zero-Copy Flattening with np.ravel() vs .flatten()
**💡 What is it?** `arr.flatten()` ALWAYS creates a deep copy; `np.ravel(arr)` creates a view whenever possible!
**🤔 Why use it?** Faster 1D flattening without wasting RAM.


```python
feature_grid_2d = np.array([[1, 2], [3, 4]])
flattened_view_1d = np.ravel(feature_grid_2d)
print('Ravel 1D View:', flattened_view_1d)
print('Shares Same Underlying Memory Buffer?:', flattened_view_1d.base is feature_grid_2d)
```
