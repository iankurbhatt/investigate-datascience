## 📁 Section 1: Understanding Memory Strides


### 🔹 1.1: Inspecting Byte Strides
**💡 What is it?** `.strides` tells NumPy how many bytes to jump in RAM to move 1 step in each dimension.
**🤔 Why use it?** Understanding low-level pointer arithmetic and contiguous memory layouts.


```python
grid_tensor = np.zeros((3, 4), dtype=np.int64)  # int64 = 8 bytes per slot
print('Grid Shape  :', grid_tensor.shape)
print('Byte Strides:', grid_tensor.strides)  # (32, 8) -> 32 bytes/row, 8 bytes/col
```

## 📁 Section 2: Sliding Window Views (Zero RAM Duplication)


### 🔹 2.1: 1D Rolling Window with sliding_window_view()
**💡 What is it?** Creates overlapping rolling sub-windows of size $W$.
**🤔 Why use it?** Blazing fast rolling moving averages, volatility bands.


```python
from numpy.lib.stride_tricks import sliding_window_view

daily_close_prices = np.array([10.0, 20.0, 30.0, 40.0, 50.0, 60.0])
rolling_price_windows = sliding_window_view(daily_close_prices, window_shape=3)
print('Rolling 3-Day Windows:\n', rolling_price_windows)
print('3-Day Moving Averages:', rolling_price_windows.mean(axis=1))
```

### 🔹 2.2: 2D Sliding Windows (Image Convolution Patches)
**💡 What is it?** Extracts moving $2D$ patches across a matrix without making copies.
**🤔 Why use it?** Computer vision convolutional filters, 2D spatial smoothing.


```python
from numpy.lib.stride_tricks import sliding_window_view

sensor_heat_map = np.arange(16).reshape(4, 4)
sliding_heat_patches = sliding_window_view(sensor_heat_map, window_shape=(2, 2))
print('Heat Map Matrix (4x4):\n', sensor_heat_map)
print('2x2 Patches Shape    :', sliding_heat_patches.shape)
```

## 📁 Section 3: Rolling Extrema Computation


### 🔹 3.1: Rolling Peak Prices via Max Aggregation
**💡 What is it?** Aggregating over the trailing window dimension.
**🤔 Why use it?** Calculating rolling drawdowns and Donchian channel breakout bounds.


```python
from numpy.lib.stride_tricks import sliding_window_view

stock_price_stream = np.array([100, 105, 95, 110, 108, 120])
rolling_peak_prices = sliding_window_view(stock_price_stream, window_shape=3).max(axis=1)
print('Rolling 3-Day Peak Prices:', rolling_peak_prices)
```
