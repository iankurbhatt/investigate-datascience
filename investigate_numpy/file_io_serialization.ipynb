## 📁 Section 1: Binary Serialization (.npy and .npz)


### 🔹 1.1: Saving and Loading Single Array (.npy)
**💡 What is it?** `np.save()` and `np.load()` store array structure and exact dtypes directly in binary format.
**🤔 Why use it?** Instant persistence without float precision loss.


```python
import os

simulated_matrix = np.random.default_rng(42).normal(size=(5, 3))
np.save('scratch_matrix.npy', simulated_matrix)
loaded_matrix = np.load('scratch_matrix.npy')
print('Loaded successfully, equal?:', np.allclose(simulated_matrix, loaded_matrix))
os.remove('scratch_matrix.npy')
```

### 🔹 1.2: Saving Multiple Arrays in Compressed Archive (.npz)
**💡 What is it?** `np.savez_compressed()` bundles multiple named arrays with zip compression.
**🤔 Why use it?** Saving machine learning train/validation/test splits together.


```python
import os

training_features = np.ones((100, 10))
training_labels = np.zeros(100)
np.savez_compressed('model_bundle.npz', X=training_features, y=training_labels)
archive_bundle = np.load('model_bundle.npz')
print('Archive Keys     :', archive_bundle.files)
print('Features X Shape :', archive_bundle['X'].shape)
os.remove('model_bundle.npz')
```

## 📁 Section 2: Text / CSV File I/O


### 🔹 2.1: Exporting & Importing CSV (savetxt & loadtxt)
**💡 What is it?** Reading and writing delimited text files.
**🤔 Why use it?** Sharing data with Excel or legacy systems.


```python
import os

pricing_data = np.array([[1.50, 2.50], [3.50, 4.50]])
np.savetxt('pricing_data.csv', pricing_data, delimiter=',', fmt='%.2f')
loaded_pricing_csv = np.loadtxt('pricing_data.csv', delimiter=',')
print('Loaded Pricing from CSV:\n', loaded_pricing_csv)
os.remove('pricing_data.csv')
```

## 📁 Section 3: Memory-Mapped Files for Huge Datasets (np.memmap)


### 🔹 3.1: Out-of-Core Processing with np.memmap()
**💡 What is it?** Treats a disk file as an in-memory NumPy array without loading it all into RAM.
**🤔 Why use it?** Processing datasets larger than physical RAM (e.g. 100GB).


```python
import os

disk_memory_map = np.memmap('large_disk_array.dat', dtype='float32', mode='w+', shape=(1000, 1000))
disk_memory_map[0, :5] = [10.0, 20.0, 30.0, 40.0, 50.0]  # Direct disk write!
disk_memory_map.flush()
print('First 5 Items on Disk Buffer:', disk_memory_map[0, :5])
del disk_memory_map
os.remove('large_disk_array.dat')
```
