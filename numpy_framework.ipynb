```python
import numpy as np
print(f"🚀 NumPy {np.__version__} Ready!")
```

---
# 🔎 STEP 0: SCENARIO DECODING & RECONNAISSANCE (The Groundwork)
> **Core Idea:** 90% of numerical errors happen because candidates rush into matrix multiplication or indexing before understanding the coordinate system or units.

### 🔹 Substep 0.1: Read the Scenario 10 Times
* **What it means:** Read through the stakeholder brief at least **10 times** until the physical or financial process (portfolio variance, latency spike, loss function) is clear.
* **Why it matters:** Prevents missing critical constraints (e.g. "annualize daily variance by multiplying by $\sqrt{252}$", "zero-out negative balances").

### 🔹 Substep 0.2: Inspect Raw Dimensions and Types

```python
# Step 0 Code Reconnaissance
user_feature_matrix = np.array([
    [25.0, 4500.0, 720.0],
    [34.0, 12000.0, 680.0],
    [42.0, 85000.0, 750.0]
], dtype=np.float64)

print("Shape (Rows x Cols) :", user_feature_matrix.shape)
print("Data Type           :", user_feature_matrix.dtype)
print("Byte Strides        :", user_feature_matrix.strides)
print("Memory Contiguity   :", user_feature_matrix.flags.c_contiguous)
```

---
# 🎯 STEP 1: TARGET OUTPUT CONTRACT (The Destination)
> **Core Idea:** Always define the output tensor dimensions, data type, and mathematical bounds before transforming data.

### 🔹 Substep 1.1: Define Output Contract Schema

```python
# Step 1 Contract Definition
# Objective: Compute Mean-Centered Standardized Feature Matrix (Z-Score)
# Target Output Contract: Shape (3, 3), dtype float64, Column Means strictly ~ 0.0
target_shape = (3, 3)
target_dtype = np.float64
print(f"Target Contract Expected: Shape {target_shape}, Type {target_dtype}")
```

---
# 🔬 STEP 2: RAW ARRAY AUDIT & MEMORY GAPS
> **Core Idea:** Audit the input matrices for missing values, infinite values, and memory layout issues.

```python
# Step 2 Gap Audit
has_nans = np.isnan(user_feature_matrix).any()
has_infs = np.isinf(user_feature_matrix).any()
print(f"Raw Matrix Audit -> Has NaNs?: {has_nans}, Has Infs?: {has_infs}")
```

---
# ⚙️ STEP 3: BACKWARD TRANSFORMATION PIPELINE
> **Core Idea:** Build clean, vectorized transformations without Python for-loops using the 6 Master Blueprints.

### 🔹 Blueprint: Axis Aggregations + Broadcasting Pipeline

```python
# Step 3 Vectorized Execution Pipeline
# 1. Calculate Feature Means along axis=0
feature_means = user_feature_matrix.mean(axis=0)

# 2. Calculate Feature Standard Deviations along axis=0
feature_stds = user_feature_matrix.std(axis=0)

# 3. Broadcast subtraction and division
standardized_feature_matrix = (user_feature_matrix - feature_means) / feature_stds
print("Standardized Feature Matrix (Z-Scores):\n", standardized_feature_matrix)
```

---
# 🛡️ STEP 4: SANITY AUDIT & NUMERICAL ASSERTIONS
> **Core Idea:** Verify output shape, finiteness, and mathematical invariants automatically.

```python
# Step 4 Automated Sanity Assertions
assert standardized_feature_matrix.shape == target_shape, "Shape mismatch!"
assert standardized_feature_matrix.dtype == target_dtype, "Dtype mismatch!"
assert np.all(np.isfinite(standardized_feature_matrix)), "Found NaN/Inf values!"

# Verify column means are approximately 0.0
computed_col_means = standardized_feature_matrix.mean(axis=0)
assert np.allclose(computed_col_means, [0.0, 0.0, 0.0], atol=1e-7), "Means are not centered!"

print("✅ ALL NUMERICAL SANITY ASSERTIONS PASSED PERFECTLY!")
```
