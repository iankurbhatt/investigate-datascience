## 🟢 Question 1 (Solved): Top 5 High-Net-Worth US Customers

### 🎯 Target Output
Produce a clean table of the **Top 5 High-Net-Worth Verified Customers in the US** sorted by `account_balance` (descending).

**Required Output Specifications:**
- **Shape:** `(5 rows, 5 columns)`
- **Columns:** `customer_id`, `full_name`, `annual_income`, `account_balance`, `credit_score`
- **Filtering Conditions:**
  1. `country_code == 'US'`
  2. `kyc_status == 'Verified'`
  3. `credit_score >= 700` (missing credit scores treated as `0`)

````python
# Solution for Question 1
df_cust = pd.read_csv('data/customers.csv')

# 1. Clean whitespace and coerce numeric types
df_cust['country_code'] = df_cust['country_code'].astype(str).str.strip()
df_cust['kyc_status'] = df_cust['kyc_status'].astype(str).str.strip()
df_cust['credit_score'] = pd.to_numeric(df_cust['credit_score'], errors='coerce').fillna(0)
df_cust['full_name'] = df_cust['first_name'].astype(str).str.strip() + ' ' + df_cust['last_name'].astype(str).str.strip()

# 2. Apply boolean filter
mask = (
    (df_cust['country_code'] == 'US') &
    (df_cust['kyc_status'] == 'Verified') &
    (df_cust['credit_score'] >= 700)
)

# 3. Sort descending and pick top 5
q1_result = (
    df_cust[mask]
    .sort_values(by='account_balance', ascending=False)
    .head(5)
    [['customer_id', 'full_name', 'annual_income', 'account_balance', 'credit_score']]
)

q1_result
````

## 🟢 Question 2 (Solved): Low Credit Score Flagging

### 🎯 Target Output
Find all customers whose `credit_score` is below **580** (or missing/NaN) and have an `account_balance` greater than **0**.

**Required Specifications:**
- **Columns to keep:** `customer_id`, `first_name`, `last_name`, `country_code`, `credit_score`, `account_balance`
- **Sorting:** Sort by `account_balance` descending.
- **Display:** Display the first 10 rows (`.head(10)`).

````python
# Solution for Question 2
df_cust = pd.read_csv('data/customers.csv')

# 1. Clean numeric credit_score (coerce missing/NaN to 0) & account_balance
df_cust['credit_score'] = pd.to_numeric(df_cust['credit_score'], errors='coerce').fillna(0)
df_cust['account_balance'] = pd.to_numeric(df_cust['account_balance'], errors='coerce').fillna(0)
df_cust['country_code'] = df_cust['country_code'].astype(str).str.strip()

# 2. Filter: credit_score < 580 and account_balance > 0
mask = (df_cust['credit_score'] < 580) & (df_cust['account_balance'] > 0)

# 3. Sort by account_balance descending and select top 10
q2_result = (
    df_cust[mask]
    .sort_values(by='account_balance', ascending=False)
    .head(10)
    [['customer_id', 'first_name', 'last_name', 'country_code', 'credit_score', 'account_balance']]
)

q2_result
````

## 🟢 Question 3 (Solved): Completed Transactions Above $500

### 🎯 Target Output
Filter `raw_transactions.csv` for transactions that satisfy the following conditions:
1. `transaction_status` is `'Completed'` (stripped whitespace).
2. `transaction_amount` is greater than or equal to `500.0`.
3. `card_type` is either `'Visa'` or `'MasterCard'`.

**Required Specifications:**
- **Columns to keep:** `transaction_id`, `customer_id`, `card_type`, `transaction_amount`, `transaction_status`
- **Sorting:** Sort by `transaction_amount` descending.
- **Display:** Display the top 5 transactions.

````python
# Solution for Question 3
df_tx = pd.read_csv('data/raw_transactions.csv')

# 1. Clean whitespace in strings and ensure transaction_amount is numeric
df_tx['transaction_amount'] = pd.to_numeric(df_tx['transaction_amount'], errors='coerce')
df_tx['transaction_status'] = df_tx['transaction_status'].astype(str).str.strip()
df_tx['card_type'] = df_tx['card_type'].astype(str).str.strip()

# 2. Apply boolean filtering
mask = (
    (df_tx['transaction_status'] == 'Completed') &
    (df_tx['transaction_amount'] >= 500.0) &
    (df_tx['card_type'].isin(['Visa', 'MasterCard']))
)

# 3. Select columns, sort descending by transaction_amount, and display top 5
q3_result = (
    df_tx[mask]
    .sort_values(by='transaction_amount', ascending=False)
    .head(5)
    [['transaction_id', 'customer_id', 'card_type', 'transaction_amount', 'transaction_status']]
)

q3_result
````

## 🟢 Question 4 (Solved): Quick Categorical Distribution & Value Counts

### 🎯 Target Output
Calculate basic distribution counts from `customers.csv`:
1. Count of customers in each `kyc_status` category.
2. Count of customers in each `account_tier` category.

````python
# Solution for Question 4
df_cust = pd.read_csv('data/customers.csv')

# 1. Clean whitespace from kyc_status and account_tier
df_cust['kyc_status'] = df_cust['kyc_status'].astype(str).str.strip()
df_cust['account_tier'] = df_cust['account_tier'].astype(str).str.strip()

# 2. Display value_counts for both columns
print('=== KYC Status Distribution ===')
print(df_cust['kyc_status'].value_counts())

print('\n=== Account Tier Distribution ===')
print(df_cust['account_tier'].value_counts())
````
